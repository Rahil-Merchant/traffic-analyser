```python
from ultralytics import YOLO

pt_model_path = r"C:\Users\swest\Desktop\traffic-capturing\traffic-reporting\models\weights\silent_contract_9082_original_bs256.pt"


def inspect_model_metadata(model_path):
    try:
        print(f"Loading model: {model_path}")
        model = YOLO(model_path)
        print("Model loaded successfully.")

        if hasattr(model, 'names'):
            class_names = model.names
            num_classes = len(class_names)

            print(f"\n--- Inferred Metadata ---")
            print(f"Number of Classes (nc): {num_classes}")
            print(f"Class Names (names):")
            if isinstance(class_names, dict): # Handles {0: 'name1', 1: 'name2'} format
                # Sort by key to ensure order if it's a dict
                sorted_names = [class_names[i] for i in sorted(class_names.keys())]
                for i, name in enumerate(sorted_names):
                    print(f"  - {name}  (ID: {i})") # Assuming keys are 0-indexed
                names_for_yaml = sorted_names
            elif isinstance(class_names, list): # Handles ['name1', 'name2'] format
                for i, name in enumerate(class_names):
                    print(f"  - {name}  (ID: {i})")
                names_for_yaml = class_names
            else:
                print("Could not determine class names format.")
                return

            print(f"\n--- Suggested data.yaml content ---")
            print(f"# Contents for your_dataset_name.yaml")
            print(f"train: ../path/to/train/images  # Optional: Update if needed")
            print(f"val: ../path/to/val/images    # Optional: Update if needed")
            print(f"# test: ../path/to/test/images # Optional")
            print(f"")
            print(f"nc: {num_classes}")
            print(f"names:")
            for name in names_for_yaml:
                print(f"  - '{name}'")

        else:
            print("Could not find 'names' attribute in the loaded model. Metadata might be missing or model type unsupported for this inspection.")

    except Exception as e:
        print(f"An error occurred: {e}")
        import traceback
        traceback.print_exc()

if __name__ == "__main__":
    inspect_model_metadata(pt_model_path)
```
